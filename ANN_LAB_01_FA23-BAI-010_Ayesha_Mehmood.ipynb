## Task 7: Loops with Break and Continue

Topics Covered: Flow Control (for / while / break / continue)

a) Use a for loop to print all numbers from 1 to 30, but skip every multiple of 4 using continue.

b) Use a for loop to print numbers from 1 to 50, and stop the loop as soon as a number greater than 20 is found, using break.

c) Use a while loop to print the multiplication table (1 to 10) of a number provided in a variable.

d) Write a loop that calculates and prints the sum of all even numbers between 1 and 100.


```python
print("1 to 30, skipping multiples of 4:")
for number in range(1, 31):
    if number % 4 == 0:
        continue
    print(number, end=" ")
print("\nStop before the first number greater than 20:")
for number in range(1, 51):
    if number > 20:
        break
    print(number, end=" ")
print("\nMultiplication table:")
table_number, multiplier = 7, 1
while multiplier <= 10:
    print(f"{table_number} x {multiplier} = {table_number * multiplier}")
    multiplier += 1
even_sum = 0
for number in range(2, 101, 2):
    even_sum += number
print("Sum of even numbers from 1 to 100:", even_sum)
```

## Task 9: Classes and Inheritance

Topics Covered: Classes (OOP Basics)

a) Create a class Course with an __init__ method that stores course_name and credit_hours, and a method display_info() that prints these details.

b) Create two objects of the Course class with different data and call display_info() on both.

c) Create a class LabCourse that inherits from Course and adds an extra attribute lab_hours, using super().__init__() to reuse the parent constructor.

d) Override the display_info() method in LabCourse so that it prints all the parent's details plus the lab_hours, and demonstrate it by creating one LabCourse object.


```python
class Course:
    def __init__(self, course_name, credit_hours):
        self.course_name = course_name
        self.credit_hours = credit_hours

    def display_info(self):
        print(f"Course: {self.course_name}; Credit hours: {self.credit_hours}")

course1 = Course("Artificial Neural Networks", 3)
course2 = Course("Linear Algebra", 3)
course1.display_info()
course2.display_info()

class LabCourse(Course):
    def __init__(self, course_name, credit_hours, lab_hours):
        super().__init__(course_name, credit_hours)
        self.lab_hours = lab_hours

    def display_info(self):
        print(f"Course: {self.course_name}; Credit hours: {self.credit_hours}; Lab hours: {self.lab_hours}")

ann_lab = LabCourse("Artificial Neural Networks Lab", 1, 3)
ann_lab.display_info()
```
