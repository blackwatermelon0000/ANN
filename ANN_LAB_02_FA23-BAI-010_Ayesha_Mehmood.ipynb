## Task 3: Addition & Subtraction

Create two 3×3 matrices representing the marks of 3 students in 3 subjects for Term 1 and Term 2.

Calculate the improvement (Term 2 − Term 1) for each student in each subject.

```python
term1 = np.array([[65, 70, 75], [72, 68, 80], [60, 62, 64]])
term2 = np.array([[70, 78, 82], [78, 76, 85], [74, 77, 80]])
total_marks = term1 + term2
improvement = term2 - term1
total_improvement = improvement.sum(axis=1)
best_students = np.flatnonzero(total_improvement == total_improvement.max()) + 1
print("Combined marks:\n", total_marks)
print("Improvement:\n", improvement)
print("Total improvement per student:", total_improvement)
print("Student number(s) with highest improvement:", best_students)
```

## Task 8: Random Matrices

Set a random seed of your choice, then generate a 3×3 matrix of random decimal values between 0 and 1.

Generate a 3×3 matrix of random integers between 10 and 100.

Generate a 4×4 matrix using the standard normal distribution (values can be negative).

Without changing the seed, run your random decimal generation code (part 1) twice and compare, explain in a comment why both outputs are identical.


The last instruction assumes identical outputs. That holds when the same seed resets the generator before each draw. Two consecutive draws from one generator normally differ. Both cases are shown below.


```python
seed = 42
rng = np.random.default_rng(seed)
decimal_matrix = rng.random((3, 3))
integer_matrix = rng.integers(10, 101, size=(3, 3))
normal_matrix = rng.standard_normal((4, 4))
print("Random decimals:\n", decimal_matrix)
print("Random integers (10 through 100):\n", integer_matrix)
print("Standard normal matrix:\n", normal_matrix)

# Same seed reset before each draw reproduces the same matrix.
first = np.random.default_rng(seed).random((3, 3))
second = np.random.default_rng(seed).random((3, 3))
print("First draw after resetting the seed:\n", first)
print("Second draw after resetting the same seed:\n", second)
print("Identical after seed reset:", np.array_equal(first, second))

# Consecutive draws without resetting advance the generator state.
continuing_rng = np.random.default_rng(seed)
next_a = continuing_rng.random((3, 3))
next_b = continuing_rng.random((3, 3))
print("Identical consecutive draws without reset:", np.array_equal(next_a, next_b))
# Keeping the seed value unchanged is not enough to repeat a draw.
# Recreating or reseeding the generator resets its state.
```

## Task 9: Transpose

Create a 3×5 matrix representing 3 employees' performance scores across 5 months.

Transpose the matrix and print its new shape.

Verify programmatically (using code, not by manual inspection) that transposing the matrix twice gives back the original matrix.

Create a 4×2 and a 3×2 matrix, and use transpose on one of them so that a dot product between the two becomes possible; then perform the dot product.


```python
performance = np.array([[70, 72, 75, 78, 80], [65, 68, 70, 73, 76], [80, 82, 81, 85, 88]])
transposed = performance.T
print("Original:\n", performance)
print("Transpose:\n", transposed)
print("New shape:", transposed.shape)
print("Double transpose restores original:", np.array_equal(performance.T.T, performance))
a = np.arange(1, 9).reshape(4, 2)
b = np.arange(1, 7).reshape(3, 2)
product = np.dot(a, b.T)
print("A shape:", a.shape, "B.T shape:", b.T.shape)
print("Dot product shape:", product.shape)
print("Dot product:\n", product)
# (4, 2) @ (2, 3) -> (4, 3); the inner dimensions match.
```
